# Limpieza y selección preliminar de variables

```python
# ================================================================
# FILTRADO DE VARIABLES ÚTILES
# ================================================================

# Calcular porcentaje de completitud por variable
completitud = 100 - (datos.isna().mean() * 100)

# Filtro: conservar variables con al menos 40% de datos válidos
umbral = 40
variables_utiles = completitud[completitud >= umbral].index.tolist()

print(f"Variables conservadas (≥ {umbral}% de completitud): {len(variables_utiles)}")
print("--------------------------------------------------")
print(variables_utiles)

# Crear nueva base reducida solo con esas variables
datos_filtrados = datos[variables_utiles].copy()

# Verificar tamaño
print("\nNueva base reducida:", datos_filtrados.shape)

# Vista rápida
datos_filtrados.head()
```

# Limpieza básica y revisión de tipos de datos

```python
# ================================================================
# LIMPIEZA BÁSICA Y REVISIÓN DE TIPOS
# ================================================================

# 1️⃣ Estandarizar nombres de columnas
datos_filtrados.columns = (
    datos_filtrados.columns
    .str.strip()
    .str.replace(' ', '_')
    .str.replace('¿', '')
    .str.replace('?', '')
    .str.replace('-', '_')
    .str.replace('.', '', regex=False)
    .str.upper()
)

# 2️⃣ Eliminar duplicados de columnas (las que terminan en .1 originalmente)
datos_filtrados = datos_filtrados.loc[:, ~datos_filtrados.columns.duplicated()]

# 3️⃣ Tipos de datos
print("Tipos de datos:")
print(datos_filtrados.dtypes.value_counts())

# 4️⃣ Visualizar las variables categóricas más frecuentes
cat_cols = datos_filtrados.select_dtypes(include='object').columns
num_cols = datos_filtrados.select_dtypes(exclude='object').columns

print(f"\nVariables categóricas: {len(cat_cols)}")
print(f"Variables numéricas: {len(num_cols)}")

# 5️⃣ Vista de ejemplo
datos_filtrados.head(3)
```

# Análisis Exploratorio Inicial (DATA UNDERSTANDING)

```python
# Porcentaje de valores nulos por columna
missing = datos_filtrados.isnull().mean().sort_values(ascending=False)*100
missing = missing[missing > 0]
print("\nVariables con valores nulos (%):\n")
print(missing)
```

```python
# Distribución de variables categóricas clave
categoricas_principales = [
    'SEXO','GENERO','ESTADO_CIVIL','CURSO_DE_VIDA',
    'ETNIA','ESTRATO','ZONA','LOCALIDAD','TIPO_INTERVENCIÓN',
    'POBLACIÓN_DIFERENCIAL_Y_DE_INCLUSIÓN'
]

cols_existentes = [col for col in categoricas_principales if col in datos_filtrados.columns]
n_cols = 2  # 3 graficas por fila

plt.figure(figsize=(5*n_cols, 5 * ((len(cols_existentes) + n_cols - 1) // n_cols)))

for i, col in enumerate(cols_existentes):
    plt.subplot((len(cols_existentes) + n_cols - 1) // n_cols, n_cols, i+1)
    sns.countplot(data=datos_filtrados, x=col, order=datos_filtrados[col].value_counts().index)
    plt.title(f"Distribución de {col}")
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()

plt.show()

# Gráfica de edad aparte
if 'EDAD' in datos_filtrados.columns:
    plt.figure(figsize=(6,4))
    sns.histplot(datos_filtrados['EDAD'], bins=20, kde=True)
    plt.title("Distribución de la EDAD")
    plt.xlabel("Edad")
    plt.ylabel("Frecuencia")
    plt.tight_layout()
    plt.show()
```

```python
# 6️⃣ Conteo de registros por localidad
if 'LOCALIDAD' in datos_filtrados.columns:
    plt.figure(figsize=(10,5))
    top_loc = datos_filtrados['LOCALIDAD'].value_counts().head(15)
    sns.barplot(x=top_loc.values, y=top_loc.index)
    plt.title("Top 15 Localidades con mayor número de registros")
    plt.xlabel("Cantidad de casos")
    plt.ylabel("Localidad")
    plt.tight_layout()
    plt.show()
```

```python
# ================================================================
# 🧹 BLOQUE COMPLETO DE IMPUTACIÓN DE VALORES FALTANTES (versión optimizada)
# ================================================================

from sklearn.impute import KNNImputer
import numpy as np
import pandas as pd

# 1️⃣ Separar variables por tipo
cat_cols = datos_filtrados.select_dtypes(include='object').columns.tolist()
num_cols = datos_filtrados.select_dtypes(include=['int64','float64','datetime64']).columns.tolist()

# 2️⃣ Eliminar columnas con más de 60% de faltantes
cols_to_drop = datos_filtrados.columns[datos_filtrados.isna().mean() > 0.6]
datos_filtrados = datos_filtrados.drop(columns=cols_to_drop)
print(f"Columnas eliminadas por alta proporción de NaN ({len(cols_to_drop)}):\n", list(cols_to_drop))

# 3️⃣ Imputaciones manuales de variables clave
if 'EDAD' in datos_filtrados.columns:
    datos_filtrados['EDAD'] = datos_filtrados['EDAD'].fillna(datos_filtrados['EDAD'].median())

if 'ESTRATO' in datos_filtrados.columns:
    datos_filtrados['ESTRATO'] = datos_filtrados['ESTRATO'].fillna(datos_filtrados['ESTRATO'].mode()[0])

if 'ESTRATO_SOCIOECONÓMICO' in datos_filtrados.columns:
    datos_filtrados['ESTRATO_SOCIOECONÓMICO'] = datos_filtrados['ESTRATO_SOCIOECONÓMICO'].fillna(datos_filtrados['ESTRATO_SOCIOECONÓMICO'].mode()[0])

# 4️⃣ Separar columnas de tipo fecha y numéricas reales
date_cols = [col for col in num_cols if np.issubdtype(datos_filtrados[col].dtype, np.datetime64)]
true_num_cols = [col for col in num_cols if col not in date_cols]

# 5️⃣ Imputación KNN (solo si hay al menos 2 columnas numéricas)
if len(true_num_cols) > 1:
    imputer = KNNImputer(n_neighbors=5)
    datos_filtrados[true_num_cols] = imputer.fit_transform(datos_filtrados[true_num_cols])
    print("✅ Imputación KNN aplicada correctamente.")

# 6️⃣ Reconversión de fechas
for col in date_cols:
    datos_filtrados[col] = pd.to_datetime(datos_filtrados[col], errors='coerce')

# 7️⃣ Imputación categórica (relleno con “No reporta”)
datos_filtrados[cat_cols] = datos_filtrados[cat_cols].fillna("No reporta")

# 8️⃣ Comprobación final
faltantes_final = datos_filtrados.isnull().sum().sum()
print(f"\nValores faltantes después de la imputación: {faltantes_final}")
```

```python
from scipy.stats import ks_2samp

# Ejemplo con la variable EDAD
original = datos['EDAD'].dropna()
imputada = datos_filtrados.loc[datos['EDAD'].isna(), 'EDAD']

stat, p = ks_2samp(original, imputada)
print(f"KS Test EDAD → estadístico={stat:.3f}, p-valor={p:.3f}")

if p > 0.05:
    print("✅ No hay diferencia significativa entre la distribución original e imputada.")
else:
    print("⚠️ Hay diferencia significativa: la imputación puede haber alterado la distribución.")
```

```python
import numpy as np

def mean_relative_diff(orig, imp):
    return np.mean(np.abs(orig.mean() - imp.mean()) / orig.mean()) * 100

mrd = mean_relative_diff(original, imputada)
print(f"Diferencia relativa media (EDAD): {mrd:.2f}%")
```

```python
# ==============================================================
# BLOQUE FINAL DE PREPROCESAMIENTO
# Selección + Conversión + Codificación + Normalización
# ==============================================================

# 📚 Librerías necesarias
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Copia del dataset base
df_proc = datos_filtrados.copy()

# --------------------------------------------------------------
# 1️⃣ SELECCIÓN DE VARIABLES
# --------------------------------------------------------------
cat_cols = df_proc.select_dtypes(include='object').columns.tolist()
num_cols = df_proc.select_dtypes(include=['float64', 'int64']).columns.tolist()

print(f"Variables categóricas: {len(cat_cols)}")
print(f"Variables numéricas: {len(num_cols)}")
```

```python

# --------------------------------------------------------------
# 2️⃣ CONVERTIR CATEGÓRICAS A STRING
# --------------------------------------------------------------
for col in cat_cols:
    df_proc[col] = df_proc[col].astype(str)

print("✅ Variables categóricas convertidas a string correctamente.")

# --------------------------------------------------------------
# 3️⃣ CODIFICACIÓN (OneHotEncoder)
# --------------------------------------------------------------
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
encoded = encoder.fit_transform(df_proc[cat_cols])

# Convertir a DataFrame
encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(cat_cols), index=df_proc.index)

# Unir con las numéricas originales
df_encoded = pd.concat([df_proc[num_cols], encoded_df], axis=1)
print(f"✅ Codificación completada. Nuevas columnas: {encoded_df.shape[1]}")
```

```python

# --------------------------------------------------------------
# 4️⃣ NORMALIZACIÓN
# --------------------------------------------------------------
scaler = StandardScaler()
df_encoded[num_cols] = scaler.fit_transform(df_encoded[num_cols])

print("✅ Normalización aplicada a las variables numéricas.")
print(f"Shape final del dataset: {df_encoded.shape}")

# --------------------------------------------------------------
# 5️⃣ OPCIONAL: GUARDAR ARCHIVO
# --------------------------------------------------------------
df_encoded.to_csv("datos_procesados.csv", index=False)
print("💾 Archivo 'datos_procesados.csv' guardado correctamente.")
```
