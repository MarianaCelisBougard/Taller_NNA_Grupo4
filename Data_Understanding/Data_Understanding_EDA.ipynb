```python
# ==============================================================
# CRISP-DM: DATA UNDERSTANDING (Análisis Exploratorio)
# ==============================================================

# Librerías
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración visual
pd.set_option('display.max_columns', None)
plt.style.use('seaborn-v0_8-whitegrid')

# ==============================================================
# CARGA DE DATOS
# ==============================================================
ruta = '/content/drive/MyDrive/Python/Proyecto NNA/base_datos_completa_NNA_TI_anon.xlsx'

diccionario = pd.read_excel(ruta, sheet_name="variables")
datos = pd.read_excel(ruta, sheet_name="BD")

print("Diccionario de variables:", diccionario.shape)
print("Base de datos principal:", datos.shape)

# Vista previa
datos.head()
```

```python
# ==============================================================
# CALIDAD DE DATOS - VALORES FALTANTES
# ==============================================================
# Reemplazar 99999 por NaN
datos.replace(99999, np.nan, inplace=True)

# Porcentaje de valores faltantes
faltantes = datos.isna().mean().sort_values(ascending=False) * 100
print("Variables con mayor porcentaje de valores faltantes:")
print(faltantes.head(20))

# Visualización del patrón de faltantes
plt.figure(figsize=(12,6))
sns.heatmap(datos.isna(), cbar=False, yticklabels=False)
plt.title("Mapa de calor de valores faltantes")
plt.show()
```

```python
# ==============================================================
# 3️⃣ DISTRIBUCIONES BÁSICAS
# ==============================================================
print("\nDistribución de edades:")
sns.histplot(data=datos, x='EDAD', bins=20, kde=True)
plt.title("Distribución de la variable EDAD")
plt.show()

print("\nDistribución por LOCALIDAD:")
plt.figure(figsize=(10,5))
sns.countplot(data=datos, y='LOCALIDAD', order=datos['LOCALIDAD'].value_counts().index)
plt.title("Frecuencia por Localidad")
plt.show()
```

```python
# ==============================================================
# 4️⃣ SELECCIÓN PRELIMINAR DE VARIABLES
# ==============================================================

# Umbral mínimo de completitud (≥ 40%)
completitud = datos.notna().mean() * 100
variables_conservadas = completitud[completitud >= 40].index.tolist()

print(f"\nVariables conservadas (≥40% de completitud): {len(variables_conservadas)}")
datos_filtrados = datos[variables_conservadas].copy()
print(f"Nueva base reducida: {datos_filtrados.shape}")
```

```python
# ==============================================================
# EXPORTAR RESULTADO A EXCEL
# ==============================================================

salida_excel = '/content/drive/MyDrive/Python/Proyecto NNA/datos_filtrados.xlsx'
with pd.ExcelWriter(salida_excel) as writer:
    datos_filtrados.to_excel(writer, index=False, sheet_name='Datos_filtrados')
    faltantes.to_frame(name='Porcentaje_faltantes').to_excel(writer, sheet_name='Faltantes')

print(f"💾 Archivo Excel guardado correctamente en: {salida_excel}")
```
